```python
from IPython.display import Audio, display
```

```python
# primero se cargan los audios
import librosa
import librosa.display
import matplotlib.pyplot as plt
import numpy as np

vozA, srA = librosa.load("VozA.wav", sr=None)
vozB, srB = librosa.load("VozB.wav", sr=None)
```

```python
print("Voz A")
display(Audio("VozA.wav"))
```

```python
print("Voz B")
display(Audio("VozB.wav"))
```

```python
# se compara la forma de la onda de cada una
plt.figure(figsize=(12,4))

plt.subplot(2,1,1)
librosa.display.waveshow(vozA, sr=srA)
plt.title("Voz A")

plt.subplot(2,1,2)
librosa.display.waveshow(vozB, sr=srB)
plt.title("Voz B")

plt.tight_layout()
plt.show()
```

```python
mfccA = librosa.feature.mfcc(y=vozA, sr=srA, n_mfcc=13)
mfccB = librosa.feature.mfcc(y=vozB, sr=srB, n_mfcc=13)

plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
librosa.display.specshow(mfccA, x_axis='time')
plt.title("MFCC Voz A")

plt.subplot(1,2,2)
librosa.display.specshow(mfccB, x_axis='time')
plt.title("MFCC Voz B")

plt.tight_layout()
plt.show()
```

```python
# espectogramas
D1 = librosa.amplitude_to_db(np.abs(librosa.stft(vozA)), ref=np.max)
D2 = librosa.amplitude_to_db(np.abs(librosa.stft(vozB)), ref=np.max)

plt.figure(figsize=(12,6))

plt.subplot(1,2,1)
librosa.display.specshow(D1, sr=srA, x_axis='time', y_axis='log')
plt.title("Espectrograma Voz A")

plt.subplot(1,2,2)
librosa.display.specshow(D2, sr=srB, x_axis='time', y_axis='log')
plt.title("Espectrograma Voz B")

plt.tight_layout()
plt.show()
```
