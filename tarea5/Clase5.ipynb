## CNN

- Las CNN son un tipo de red neuronal diseñada para procesar datos en forma de múltiples arrays, como imágenes.
- Se utilizan ampliamente en tareas de visión por computadora debido a su capacidad para captar características espaciales jerárquicas.

$$(f*g)(t):=\int _{-\infty }^{\infty }f(\tau )g(t-\tau )\,d\tau$$


![](https://upload.wikimedia.org/wikipedia/commons/b/b9/Convolution_of_spiky_function_with_box2.gif)

### Concepto de convolución
- La convolución es una operación matemática que combina dos conjuntos de información.
- En el contexto de las CNN, la convolución se utiliza para extraer características de la imagen a través de filtros (_kernels_).

![](https://upload.wikimedia.org/wikipedia/commons/1/19/2D_Convolution_Animation.gif)

- Reducción de parámetros comparada con redes completamente conectadas.
- Preservación de relaciones espaciales en la imagen.

![](https://editor.analyticsvidhya.com/uploads/81269Capture.PNG)


### Arquitectura de una CNN

#### Capas convolucionales
- Las capas convolucionales aplican filtros sobre la imagen de entrada para crear mapas de características.
- Cada filtro es una matriz pequeña que se desliza (convoluciona) sobre la imagen para detectar patrones como bordes, texturas, etc.

#### Capas de _pooling_
- Las capas de _pooling_ reducen la dimensionalidad de los mapas de características, manteniendo la información más relevante.
- Max pooling: selecciona el valor máximo en una ventana de la imagen.
- Average pooling: calcula el promedio de los valores en una ventana de la imagen.

#### Capas totalmente conectadas
- Estas capas conectan todas las neuronas de una capa a todas las neuronas de la siguiente capa.
- Se utilizan para combinar las características extraídas y realizar la clasificación final.

![](https://editor.analyticsvidhya.com/uploads/675071_vkQ0hXDaQv57sALXAJquxA.jpeg)

![](https://editor.analyticsvidhya.com/uploads/18707neural-networks-layers-visualization.jpg)


### Función de activación y función de pérdida

#### Funciones de activación
  - Las funciones de activación introducen no linealidad en la red, permitiendo que la red aprenda de datos complejos.

#### Función de activación ReLU
  - ReLU: $f(x) = \max(0, x)$
  - Ventajas: simple, eficiente computacionalmente, ayuda a mitigar el problema del desvanecimiento del gradiente.
  - Gráfica de la función ReLU.

#### Función de activación Sigmoid
  - Sigmoid: $f(x) = 1 / (1 + e^{-x})$
  - Utilizada en la salida para problemas de clasificación binaria.
  - Gráfica de la función Sigmoid.

#### Función de activación Softmax
  - Softmax: $f(x_i) = e^{(x_i)} / \sum(e^{(x_j)})$
  - Utilizada en la salida para problemas de clasificación multiclase.
  - Gráfica de la función Softmax aplicada a un vector.

  ![](https://www.researchgate.net/profile/Hendrik_Pfaff/publication/371831710/figure/fig3/AS:11431281170178629@1687641707821/Plot-of-the-sigmoid-ReLU-and-softmax-activation-functions.ppm)


### Implementación de CNN

Carga de dependencias

```python
import tensorflow as tf
from tensorflow.keras import datasets, layers, models
import matplotlib.pyplot as plt
```

Carga de conjunto de datos CIFAR-10

![](https://storage.googleapis.com/kaggle-datasets-images/1424838/2359346/083dc91e6e5c66971c05f2439ebc5144/dataset-cover.jpg?t=2021-06-22-09-51-53)

```python
(train_images, train_labels), (test_images, test_labels) = datasets.cifar10.load_data()
train_images, test_images = train_images / 255.0, test_images / 255.0
```

Construcción del modelo


```python
model = models.Sequential([
    layers.Conv2D( # Capa de entrada
        32, (3, 3), # Aplica 32 filtros de 3 × 3
        activation='relu', # 2. Aplica función de activación ReLU
        input_shape=(32, 32, 3) # 1. Recibe imágenes de 32 × 32 con tres colores (RGB)
    ),
    layers.MaxPooling2D((2, 2)), # Reducción de dimensionalidad a 2 × 2
    layers.Conv2D(64, (3, 3), activation='relu'), # 64 filtros de 3 × 3
    layers.MaxPooling2D((2, 2)), # Reducción de dimensionalidad a 2 × 2
    layers.Conv2D(64, (3, 3), activation='relu'), # 64 filtros de 3 × 3
    layers.Flatten(), # Conversión a vector para capa densa (totalmente conectada)
    layers.Dense(64, activation='relu'), # Capa densa de 64 neuronas
    layers.Dense(10) # Capa de salida con una neurona por cada clase
])
```

Compilación del modelo

Optimizadores a partir de gradientes

![](https://miro.medium.com/v2/resize:fit:640/format:webp/1*hUd744hDEEGx0-ypWGhrkw.gif)

Ejemplos:
- Cian: Stochastic Gradient Descent (SGD)
- Azul: Adam (Adaptive Moment Estimation)
- Magenta: Momentum
- Blanco: Adagrad (Adaptive Gradient Algorithm)
- Verde: RMSProp

![image.png](https://miro.medium.com/v2/resize:fit:720/format:webp/1*47skUygd3tWf3yB9A10QHg.gif)

```python
model.compile(
    optimizer = 'adam', # optimizador
    loss = tf.keras.losses.SparseCategoricalCrossentropy(from_logits=True),
    metrics = ['accuracy']
)
```

Entrenamiento del modelo

```python
history = model.fit(
    train_images, train_labels, epochs = 10,
    validation_data = (test_images, test_labels)
)
```

Resultados

```python
plt.plot(history.history['loss'], label='loss')
plt.plot(history.history['val_loss'], label = 'val_loss')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.ylim([0, 1])
plt.legend(loc='lower right')
plt.show()
```


## Transfer Learning


### Definición de Transfer Learning
- Transfer Learning es una técnica en aprendizaje automático donde se utiliza un modelo preentrenado en una tarea diferente a la que fue originalmente entrenado. Esta técnica es especialmente útil cuando se dispone de un conjunto de datos limitado para la tarea actual.

![Ejemplo de transfer learning](https://www.researchgate.net/profile/Mazin-Mohammed/publication/347946990/figure/fig3/AS:973645212049408@1609146516624/Example-of-transfer-learning-in-medical-images.png "Ejemplo de transfer learning")

### Ventajas del Transfer Learning
- **Menor tiempo de entrenamiento**: Los modelos preentrenados ya han aprendido características básicas y complejas, por lo que solo necesitan ajustes menores, reduciendo significativamente el tiempo de entrenamiento.
- **Menos datos requeridos**: Los modelos preentrenados han aprendido de conjuntos de datos masivos, lo que permite obtener buenos resultados incluso con conjuntos de datos más pequeños en la nueva tarea.


### Aplicaciones comunes
  - **Clasificación de imágenes**: Uso de modelos preentrenados para clasificar imágenes en diferentes categorías.
  - **Detección de objetos**: Identificación y localización de objetos dentro de una imagen.
  - **Segmentación de imágenes**: Dividir una imagen en segmentos significativos, como separar el fondo de los objetos en primer plano.
  
![Ejemplo de transfer learning](https://media.springernature.com/lw685/springer-static/image/art%3A10.1007%2Fs00521-018-3468-3/MediaObjects/521_2018_3468_Fig1_HTML.gif "Ejemplo de transfer learning")




### Modelos populares para Transfer Learning
- **VGG16**: Modelo de red profunda con 16 capas, conocido por su simplicidad y efectividad en clasificación de imágenes.
![VGG16](https://media.geeksforgeeks.org/wp-content/uploads/20200219152207/new41.jpg)

- **ResNet50**: Red residual con 50 capas, conocida por su capacidad para entrenar modelos extremadamente profundos sin problemas de desvanecimiento del gradiente.

![ResNet50](https://miro.medium.com/v2/resize:fit:720/format:webp/0*tH9evuOFqk8F41FG.png)

- **InceptionV3**: Modelo que utiliza módulos Inception para capturar información a múltiples escalas y profundidades, conocido por su eficiencia y precisión.

![ResNet50](https://www.researchgate.net/publication/349717475/figure/fig5/AS:996933934014464@1614698980419/The-architecture-of-Inception-V3-model.ppm)

- **MobileNet**: Modelo diseñado para dispositivos móviles y embebidos, optimizado para eficiencia y velocidad sin sacrificar demasiado rendimiento.

![ResNet50](https://www.researchgate.net/publication/357623745/figure/fig5/AS:1109250583593087@1641477356627/The-architecture-of-MobileNet-V1-30.jpg)

### Implementación de Transfer learning

Carga del modelo

```python
from tensorflow.keras.applications import VGG16
from tensorflow.keras import layers, models
import tensorflow_datasets as tfds

base_model = VGG16(
    weights='imagenet',
    include_top=False,
    input_shape=(150, 150, 3)
)
base_model.trainable = False
```

Construcción del nuevo modelo

```python
model = models.Sequential([
    base_model,
    layers.Flatten(),
    layers.Dense(256, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

Carga del conjunto de datos

![](https://storage.googleapis.com/tfds-data/visualization/fig/cats_vs_dogs-4.0.1.png)

```python
dataset, info = tfds.load('cats_vs_dogs', with_info=True, as_supervised=True)
train_dataset, test_dataset = dataset['train'].take(20000), dataset['train'].skip(20000)
def preprocess(image, label):
    image = tf.image.resize(image, (150, 150))
    image = image / 255.0
    return image, label
train_dataset = train_dataset \
    .map(preprocess) \
    .batch(32) \
    .prefetch(tf.data.AUTOTUNE)
test_dataset = test_dataset \
    .map(preprocess) \
    .batch(32) \
    .prefetch(tf.data.AUTOTUNE)
```

Entrenamiento del modelo

```python
history = model.fit(
    train_dataset,
    epochs=10,
    validation_data=test_dataset)
```

Resultados

```python
plt.plot(history.history['accuracy'], label='accuracy')
plt.plot(history.history['val_accuracy'], label = 'val_accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.ylim([0, 1])
plt.legend(loc='lower right')
plt.show()
```

## _Fine-tuning_

### Definición de _Fine-tuning_
-  Ajustar modelo preentrenado para mejorar su rendimiento en una tarea específica.
- Se copian capas originales, se entrena capa densa

![](https://b2633864.smushcdn.com/2633864/wp-content/uploads/2019/06/fine_tuning_keras_network_surgery.png)

- También es posible entrenar todas las capas

![](https://b2633864.smushcdn.com/2633864/wp-content/uploads/2019/06/fine_tuning_keras_freeze_unfreeze.png)

### Diferencia entre Transfer Learning y Fine-tuning

**Transfer Learning**:
- En Transfer Learning, se utilizan los pesos del modelo preentrenado y se entrenan solo las capas superiores añadidas específicamente para la nueva tarea.
- Es rápido y requiere menos datos.
- Ideal cuando los datos específicos de la nueva tarea son limitados.

![](https://substackcdn.com/image/fetch/w_1456,c_limit,f_webp,q_auto:good,fl_progressive:steep/https%3A%2F%2Fsubstack-post-media.s3.amazonaws.com%2Fpublic%2Fimages%2Fd0aefff8-522e-4502-9ab4-a1fe9d02d2d2_1060x308.gif)

**Fine-tuning**:
- En Fine-tuning, además de entrenar las capas superiores, se ajustan algunas o todas las capas del modelo base.
- Puede llevar más tiempo y requiere más datos.
- Ideal cuando se dispone de un conjunto de datos más grande y se necesita una mayor precisión.

![](https://substackcdn.com/image/fetch/f_auto,q_auto:good,fl_progressive:steep/https%3A%2F%2Fsubstack-post-media.s3.amazonaws.com%2Fpublic%2Fimages%2F9764beac-a786-4305-9a47-ec050b0ebef6_1060x308.gif)

### Estrategias de Fine-tuning

**Congelación inicial de capas y posterior descongelación selectiva**:
- Primero, se congelan todas las capas del modelo preentrenado para entrenar solo las capas superiores añadidas.
- Posteriormente, se descongelan selectivamente las capas del modelo base para ajustar finamente solo algunas partes del modelo, lo cual reduce el riesgo de sobreajuste y permite que el modelo aprenda características específicas de la nueva tarea.

**Ajuste de la tasa de aprendizaje**:
- Utilizar una tasa de aprendizaje más baja para las capas preentrenadas y una tasa de aprendizaje más alta para las nuevas capas.
- Este enfoque permite realizar ajustes sutiles en las capas preentrenadas sin afectar significativamente los pesos ya aprendidos.

### Implementación de Fine-tuning

Carga de dependencia

```python
from tensorflow.keras.applications import ResNet50
from tensorflow.keras import layers, models
```

Carga del modelo

```python
base_model = ResNet50(weights='imagenet', include_top=False, input_shape=(150, 150, 3))
base_model.trainable = True
```

Ajuste de capas

```python
for layer in base_model.layers[:-10]:
    layer.trainable = False

model = models.Sequential([
    base_model,
    layers.GlobalAveragePooling2D(),
    layers.Dense(256, activation='relu'),
    layers.Dense(1, activation='sigmoid')
])
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

Entrenamiento y evaluación

```python
train_dataset = train_dataset.map(preprocess).batch(32).prefetch(tf.data.AUTOTUNE)
test_dataset = test_dataset.map(preprocess).batch(32).prefetch(tf.data.AUTOTUNE)

history = model.fit(train_dataset, epochs=10, validation_data=test_dataset)
```

Resultados

```python
plt.plot(history.history['accuracy'], label='accuracy')
plt.plot(history.history['val_accuracy'], label='val_accuracy')
plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.ylim([0, 1])
plt.legend(loc='lower right')
plt.show()
```
